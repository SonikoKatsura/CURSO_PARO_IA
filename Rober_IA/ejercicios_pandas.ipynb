### 🧪 Nivel 3: Limpieza de datos

11. **Cuenta cuántos valores nulos hay en cada columna.**

12. **Elimina todas las filas donde falte el valor de "Age".**

13. **Rellena los valores faltantes en "Embarked" con el valor más frecuente.**

14. **Crea una nueva columna "EdadRedondeada" que contenga la edad redondeada.**

15. **Reemplaza los valores de la columna "Sex" por "Hombre" y "Mujer".**

---

````python
# 11. Cuenta cuántos valores nulos hay en cada columna.
print("11. Número de valores nulos por columna:\n")
df.isnull().sum()
````

````python
# 12. Elimina todas las filas donde falte el valor de "Age".

# Mostrar el número de filas antes de la eliminación
filas_antes = df.shape[0]
print("\n12. Eliminando filas con valores nulos en 'Age':")
print(f"Número de filas en el DataFrame original: {filas_antes}")
print("\nPrimeras filas del DataFrame original (con posibles nulos en 'Age'):")
print(df[['Age']].sample(10))

# Eliminar las filas con valores nulos en "Age"
df_sin_nulos_edad = df.dropna(subset=['Age']).copy()

# Mostrar el número de filas después de la eliminación
filas_despues = df_sin_nulos_edad.shape[0]
print(f"\nNúmero de filas después de eliminar los nulos en 'Age': {filas_despues}")
print("\nPrimeras filas del DataFrame después de eliminar los nulos en 'Age':")
print(df_sin_nulos_edad[['Age']].sample(10))
````

````python
# 13. Rellena los valores faltantes en "Embarked" con el valor más frecuente.

# Mostrar número de nulos antes
print("\n13. Rellenando valores nulos en 'Embarked':")
print(f"Número de nulos en 'Embarked' antes del relleno: {df_sin_nulos_edad['Embarked'].isnull().sum()}")
print("\nPrimeras filas del DataFrame antes del relleno de 'Embarked':")
print(df_sin_nulos_edad[['Embarked']].sample(10))

# Calcular el valor más frecuente
valor_mas_frecuente_embarked = df_sin_nulos_edad['Embarked'].mode()[0]

# Rellenar los valores faltantes (manera recomendada)
df_sin_nulos_edad['Embarked'] = df_sin_nulos_edad['Embarked'].fillna(valor_mas_frecuente_embarked)

# Mostrar número de nulos después
print(f"\nNúmero de nulos en 'Embarked' después del relleno: {df_sin_nulos_edad['Embarked'].isnull().sum()}")
print("\nPrimeras filas del DataFrame después del relleno de 'Embarked':")
print(df_sin_nulos_edad[['Embarked']].sample(10))
````

````python
# 14. Crea una nueva columna "EdadRedondeada" que contenga la edad redondeada.
df_sin_nulos_edad['EdadRedondeada'] = df_sin_nulos_edad['Age'].round().astype(int)
print("\n14. DataFrame con la nueva columna 'EdadRedondeada':")
df_sin_nulos_edad[['Age', 'EdadRedondeada']].sample(10)
````

````python
# 15. Reemplaza los valores de la columna "Sex" por "Hombre" y "Mujer".
df_sin_nulos_edad['Sex'].replace({'male': 'Hombre', 'female': 'Mujer'}, inplace=True)
print("\n15. DataFrame con los valores de 'Sex' reemplazados (muestra aleatoria):")
df_sin_nulos_edad[['PassengerId', 'Name', 'Sex']].sample(10)
````

### 🧪 Nivel 4: Agrupamiento y análisis

16. **Cuenta cuántos pasajeros hay por clase (Pclass).**

17. **Calcula la media de edad por clase de pasajero.**

18. **Agrupa por "Sex" y "Pclass" y muestra la tarifa media pagada ("Fare").**

19. **Agrupa por "Embarked" y muestra cuántos sobrevivieron en cada puerto de embarque.**

20. **Crea una tabla pivote que muestre la media de edad según el sexo y la clase ("Pclass").**

---

````python
# 16. Cuenta cuántos pasajeros hay por clase (Pclass).
print("\n16. Número de pasajeros por clase (Pclass):\n")
df_sin_nulos_edad['Pclass'].value_counts()
````

````python
# 17. Calcula la media de edad por clase de pasajero.
print("\n17. Media de edad por clase de pasajero:\n")
df_sin_nulos_edad.groupby('Pclass')['Age'].mean()
````

````python
# 18. Agrupa por "Sex" y "Pclass" y muestra la tarifa media pagada ("Fare").
print("\n18. Tarifa media pagada por sexo y clase:\n")
df_sin_nulos_edad.groupby(['Sex', 'Pclass'])['Fare'].mean()
````

````python
# 19. Agrupa por "Embarked" y muestra cuántos sobrevivieron en cada puerto de embarque.
print("\n19. Número de sobrevivientes por puerto de embarque:\n")
df_sin_nulos_edad.groupby('Embarked')['Survived'].sum()
````

````python
# 20. Crea una tabla pivote que muestre la media de edad según el sexo y la clase ("Pclass").
tabla_pivot_edad = df_sin_nulos_edad.pivot_table(values='Age', index='Sex', columns='Pclass')
print("\n20. Tabla pivote de la media de edad por sexo y clase:\n")
tabla_pivot_edad
````
