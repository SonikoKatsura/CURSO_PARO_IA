f) Contestar a las siguientes preguntas:

¿Cuál es la media de los cargos médicos?

¿Y la desviación típica o estándar (std)?
 Podrías decir con tus palabras si los datos están mas concentrados o al contrario mas distribuidos respecto a la media?

Valores máximo y mínimo de hijos (Children)

¿Cuál es la mediana del tomador del seguro?

```python
print("f) Análisis estadístico básico:")

# ¿Cuál es la media de los cargos médicos?
media_cargos = df_insurance['charges'].mean()
print(f"\n¿Cuál es la media de los cargos médicos?\n {media_cargos:.2f}")

# ¿Y la desviación típica o estándar (std)? Podrías decir con tus palabras si los datos están mas concentrados o al contrario mas distribuidos respecto a la media?
std_cargos = df_insurance['charges'].std()
print(f"¿Y la desviación típica o estándar (std)?\n {std_cargos:.2f}")
print("\nLa desviación estándar de los cargos es alta ({:.2f}) comparada con la media ({:.2f}),\nasí que los precios de las facturas están bastante esparcidos, no amontonados cerca del promedio.".format(std_cargos, media_cargos))

# Valores máximo y mínimo de hijos (Children)
max_hijos = df_insurance['children'].max()
min_hijos = df_insurance['children'].min()
print(f"\nValores máximo y mínimo de hijos (Children):\n Máximo = {max_hijos} \n Mínimo = {min_hijos}")

# ¿Cuál es la mediana del tomador del seguro?
mediana_edad = df_insurance['age'].median()
print(f"\nLa mediana de edad del tomador del seguro es: {mediana_edad} años.")
```

Determina cuál es el valor mas frecuente de la columna "smoker" y con que frecuencia. ¿Dirías que es un valor factible (posible, real)?

```python
valor_frecuente_smoker = df_insurance['smoker'].mode()[0]
frecuencia_smoker = df_insurance['smoker'].value_counts()[valor_frecuente_smoker]

print(f"El valor más frecuente en la columna 'smoker' es '{valor_frecuente_smoker}' con una frecuencia de {frecuencia_smoker} individuos.")

conteo_fumadores = df_insurance['smoker'].value_counts()
proporcion_fumadores = df_insurance['smoker'].value_counts(normalize=True) * 100

print("\nDistribución de fumadores:")
print(conteo_fumadores)
print("\nProporción de fumadores (en porcentaje):")
print(proporcion_fumadores.round(2))

print("\nPensando en las tasas de tabaquismo en Estados Unidos en general, históricamente han sido más altas,\npero han ido disminuyendo con el tiempo gracias a las campañas de concienciación y las regulaciones.\nSin embargo, todavía hay un porcentaje significativo de la población adulta que fuma.\nAsí que, los porcentajes que obtengamos de nuestra tabla deberían ser posibles dentro del contexto de la población estadounidense.\nSi vemos una proporción muy alta de fumadores o muy baja en comparación con lo que sabemos de las tendencias generales en EEUU,\npodríamos empezar a preguntarnos si nuestra muestra es representativa de toda la población.\nPor ejemplo, si en nuestra tabla sale que fuma el 50% de la gente, eso podría ser un poco alto para las tasas actuales a nivel nacional,\npero podría ser posible en ciertos grupos de edad o regiones específicas. Si sale que fuma solo el 5%, sería bastante bajo para la media general.\nEn resumen, lo que veamos en la tabla será un reflejo de esta muestra en particular de gente con seguros en EEUU,\ny debería ser factible dentro de ese contexto, aunque podría no ser exactamente igual a la tasa nacional general.")
```
