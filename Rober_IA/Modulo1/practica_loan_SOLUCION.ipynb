## AÑADIR CAMPO CALCULADO "Total_Problems" como suma de "Number of credit problems" y "Bankruptcies"

```python
df_loan["Total_Problems"] = df_loan["Number_of_Credit_Problems"] + df_loan["Bankruptcies"]

df_loan.head()
```

## CODIFICAR LA VARIABLE OBJETIVO

```python
df_loan.Loan_Status.unique()
```

Vamos a codificarla como:

- "Charged Off": 0
- "Fully Paid": 1

```python
df_loan.Loan_Status.replace({'Charged Off': 0, 'Fully Paid': 1}, inplace=True)

df_loan.Loan_Status.unique()
```

```python
df_loan.columns
```

## NOS HACEMOS ALGUNAS PREGUNTAS.

### **-¿Hay relación entre el Loan Status (target) y Annual Income?**

```python
df_loan.groupby(by=['Loan_Status'])['Annual_Income'].agg(['mean','min', 'max'])
```

En los impagados se observan cifras mayores de Ingreso, con lo cual la relación con Status no es tan fuerte (debe probablemente de depender también de otras variables)

### **-¿Hay relación entre el Loan Status (target) y el Current Loan Amount?**

```python
df_loan.groupby(by=['Loan_Status'])['Current_Loan_Amount'].agg(['mean','min', 'max'])
```

La media para los pagados supera en 2 ordenes de magnitud, sin embargo puede ser debido a la presencia de outliers (999999999---)

### **- ¿El tipo de propiedad influye en si se paga o no un préstamo?**

```python
df_loan.groupby(by=['Home_Ownership', 'Loan_Status'])['Loan_Status'].agg('count')
```

En los grupos mayoritarios, la relación parece ser la misma: los impagados son aproximadamente un tercio de los pagados
